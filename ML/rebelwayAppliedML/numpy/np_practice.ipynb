```python
import numpy as np
```

```python
# 深拷贝与浅拷贝问题
a =np.arange(0,15)
a
```

```python
a[1:3]
```

```python
a[1:3] = 9
a
```

```python
# 浅拷贝
b = a[1:3]
b[:] = 999
b
a
```

```python
# 深拷贝
c = a[1:3].copy()
print(c)
print(a)
c[:] = 111
print(a)
```
